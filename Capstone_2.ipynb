### Menguji menggunakan statistik inferensial (goodness of fit test) pada profit apakah berdistribusi normal atau tidak

```python
# Secara visualisasi, banyak outlier
# coba kita uji statistik apakah data profitnya berdistribusi normal/tidak
#h0 =  data berdistribusi normal, pvalue = 0.05

from scipy.stats import goodness_of_fit
import scipy

result = goodness_of_fit(
    dist = scipy.stats.norm, #Uji statistik untuk menguji apakah data berdistribusi normal
    data = df['Profit'],
)
print(result)
```

Dari uji statistik 

success :  True (Proses fitting berhasil dilakukan)
pvalues : 0.0001 lebih kecil dari 0.05 maka (menolak hipotesis H0)

hipotesis 
H0 = data profit berdistribusi normal
H1 = data profit tidak berdistribusi normal

Karena p-value = 0.0001 < 0.05, maka:
Kesimpulannya: Data Profit tidak berdistribusi normal secara statistik.

```python
# Experiment : uji distribusi data dari data yang sudah dibersihkan outliernya
# step 1 : hapus outlier dari data profit

df[(df['Profit']>0) & (df['Profit']<30)]['Profit'].describe()
```

bisa kita lihat diatas kalau banyak skali data yang terbuang disini data awal 9.994 dan data setelah kita bersihkan menjadi 5.595 data
artinya kita membuang 4.399 data hampir dari setengah dari keseluruhan data yang terbuang dimana ini tidak baik untuk analisis


setelah menghapus outlier kita visualisasikan dengan menggunakan boxplot

```python
sns.boxplot(
    data=df[(df['Profit']>0) & (df['Profit']<30)]['Profit']
)
```

```python

result = goodness_of_fit(
    dist = scipy.stats.norm, #Uji statistik untuk menguji apakah data berdistribusi normal
    data=df[(df['Profit']>0) & (df['Profit']<30)]['Profit'],
)
print(result)
```

pvalue = 0.0001 (lebih kecil dari 0.05) 
Kesimpulannya: Data Profit yang sudah clean tidak berdistribusi normal secara statistik.

```python
#experimental menggunakan shapiro
from scipy.stats import shapiro

stat, p = shapiro(df['Profit'])
print("Shapiro-Wilk Test:", stat, p)
```

dari uji Goodness of Fit Test diatas maka dapat disimpulkan bahwa data Sales , Discount , Quantity dan Profit tidak berdistribusi normal
ketika data numerikal tidak berdistribusi normal maka hal selanjutnya yang kita lakukan adalah menggunakan metode non-parametric

### Mengidentifikasi Produk dengan total Profit dari tertinggi ke terendah

```python
product_profit = df.groupby('Product')['Profit'].sum().sort_values(ascending=False)
print(product_profit)
```

Visualisasi product dengan total profit tertinggi hingga terendah menggunakan Bar chart
mengapa menggunakan barchart : barchart sangat efektif untuk membandingkan nilai antar kategori, dalam konteks ini kategori adalah product dan nilainya adalah total profit dari produk

```python
product_profit_total = product_profit.reset_index()
plt.figure(figsize=(13, 5))
sns.barplot(
    data = product_profit_total,
    x = "Product",
    y = "Profit",
)
plt.xticks(rotation = 90)
plt.title('Total Profit per Product')
plt.xlabel('Product')
plt.ylabel('Profit')
plt.show()
```

bisa kita lihat disini perbedaan yang sangat signifikan , kita bisa lihat juga ada masalah pada marketing suite profitnya negatif(merugi) untuk memastikan apakah

## Uji Perbedaan Profit antar Produk Menggunakan Non-Parametric (Kruskal-Wallis Test)

```python
from scipy.stats import kruskal 

kruskal(
    df[df['Product']=='Alchemy']['Profit'],
    df[df['Product']=='Site Analytics']['Profit'],
    df[df['Product']=='Data Smasher']['Profit'],
    df[df['Product']=='Support']['Profit'],
    df[df['Product']=='FinanceHub']['Profit'],
    df[df['Product']=='Marketing Suite - Gold']['Profit'],
    df[df['Product']=='OneView']['Profit'],
    df[df['Product']=='SaaS Connector Pack']['Profit'],
    df[df['Product']=='ContactMatcher']['Profit'],
    df[df['Product']=='SaaS Connector Pack - Gold']['Profit'],
    df[df['Product']=='ChatBot Plugin']['Profit'],
    df[df['Product']=='Big Ol Database']['Profit'],
    df[df['Product']=='Storage']['Profit'],
    df[df['Product']=='Marketing Suite']['Profit']
)
```

Dari hasil diatas kita bisa lihat bahwa
Statistik uji H: 1086.91 Semakin besar nilainya, semakin kuat bukti menolak H0
p-value: 3.71e-224 (sangat kecil, hampir nol)

Ada perbedaan signifikan dalam profit setidaknya antara satu produk dengan produk lainnya.

```python
subregion_profit = df.groupby('Subregion')['Profit'].sum().sort_values(ascending=False)
print(subregion_profit)
```

## Pengaruh Diskon terhadap Profit (Spearman)

```python
from scipy import stats
res = stats.spearmanr(df['Discount'], df['Profit'])
print(res)
```

Dari visualisasi dan juga uji spearman correlation antara kolom discount dan profit bisa kita simpulkan :

- Kita bisa lihat nilai statistic -0.5433 artinya korelasi negatif yang cukup kuat antara diskon dan profit
- pvalue < 0.05 ini menunjukkan bahwa hubungan negatif tersebut bukan karena kebetulan, melainkan memang pola yang kuat secara statistik.
- Semakin besar diskon, semakin kecil profit
- Diskon tinggi berisiko rugi dari discount 30% - 80% cenderung titik profit di minus
